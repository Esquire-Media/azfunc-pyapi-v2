# Set your options

query_type options are:

* "distinct_mappings"
  * gets just the deduped pairs of [original name : mapped name] across all batches from that tenant
* "batch_metadata_mappings_simple"
  * gets all mapped pairs for each of the batches for that tenant
* "batch_metdata_mappings"
  * same as above, but also includes all of the metadata for that sales batch (uploader, source_file, etc)

```python
tenant_id = "cm6fe3xnw0001dupu03t376ue"

query_type = "distinct_mappings"
# query_type = "batch_metadata_mappings_simple"
# query_type = "batch_metdata_mappings"
```

# Run and view the results

```python
results = get_sales_batch_mappings(
    tenant_id=tenant_id,
    query_type=query_type
    )
```
